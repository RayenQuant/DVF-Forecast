# 02 — Feature engineering

Construction des variables qui alimenteront le modèle XGBoost.
Ce notebook reproduit interactivement ce que fait `src.features.build_features`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from src.config import CLEAN_PARQUET
from src.features import (
    add_temporal_features,
    add_geographic_features,
    add_typological_features,
    encode_categoricals,
)

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_parquet(CLEAN_PARQUET)
df = add_temporal_features(df)
df = add_geographic_features(df)
df = add_typological_features(df)
df, encoders = encode_categoricals(df)
df.head()
```

## Corrélation avec la cible

```python
numeric = df.select_dtypes('number')
corr = numeric.corr()['log_prix_m2'].sort_values(ascending=False)
corr
```

```python
plt.figure(figsize=(8, 6))
corr.drop('log_prix_m2').plot.barh()
plt.title("Corrélations avec log(prix_m2)")
plt.show()
```

Les médianes géographiques (commune, département) sont sans surprise les variables
les plus corrélées : elles encapsulent la prime de localisation. Surface et
encodage du type de bien apportent un signal complémentaire.
